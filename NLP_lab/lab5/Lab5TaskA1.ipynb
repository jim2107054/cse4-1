# Lab Task A1: Token-Level Sequence Labeling (Token Classifier)

### Task Description
Convert the Transformer from a sequence-level classifier into a **Token-Level Sequence Labeler** (e.g., for Part-of-Speech tagging or Named Entity Recognition). Instead of pooling sequence outputs into a single vector, classify every individual token in the sequence.

### Instructions
1. Implement `TokenClassifierTransformer.__init__()` and `TokenClassifierTransformer.forward()`.
2. Omit pooling. Pass all sequence states directly into the linear classifier to output logits for every token position.

```python
import torch
import torch.nn as nn
import math

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        return x + self.pe[:, :x.size(1), :]

class TokenClassifierTransformer(nn.Module):
    def __init__(self, vocab_size, num_tags=4, d_model=64, nhead=2, num_layers=2, max_len=32):
        super().__init__()
        self.d_model = d_model
        
        # =========================================================================
        # TODO: Implement model layers for token-level tagging
        # 1. Embedding layer (padding_idx=0)
        # 2. PositionalEncoding instance
        # 3. TransformerEncoderLayer and TransformerEncoder (batch_first=True)
        # 4. Linear classifier head (d_model -> num_tags) applied to every token
        # =========================================================================
        
        # YOUR CODE HERE
        self.embedding = nn.Embedding(vocab_size, d_model, padding_idx=0)
        self.pos_encoder = PositionalEncoding(d_model, max_len)
        encoder_layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=nhead, batch_first=True)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc = nn.Linear(d_model, num_tags)

    def forward(self, x, padding_mask=None):
        # =========================================================================
        # TODO: Implement forward pass
        # 1. Embed x and scale by sqrt(d_model)
        # 2. Add positional encoding
        # 3. Pass through transformer encoder with padding_mask
        # 4. Project every token's hidden state directly through classifier
        # 5. Return tensor of shape (Batch_Size, Sequence_Length, num_tags)
        # =========================================================================
        
        # YOUR CODE HERE
        embed = self.embedding(x) * math.sqrt(self.d_model)
        embed = self.pos_encoder(embed)
        out = self.transformer_encoder(embed, src_key_padding_mask=padding_mask)
        logits = self.fc(out)
        return logits

# =============================================================================
# Automated Verification Suite (Do not modify)
# =============================================================================
def test_task_A1():
    torch.manual_seed(42)
    dummy_vocab_size = 80
    batch_size, seq_len, num_tags = 4, 12, 5
    
    dummy_input = torch.randint(1, dummy_vocab_size, (batch_size, seq_len)).to(device)
    dummy_mask = (dummy_input == 0).to(device)
    
    model = TokenClassifierTransformer(vocab_size=dummy_vocab_size, num_tags=num_tags, d_model=64).to(device)
    
    try:
        logits = model(dummy_input, padding_mask=dummy_mask)
        assert logits is not None, "Forward pass returned None."
        assert logits.shape == torch.Size([batch_size, seq_len, num_tags]), \
            f"Expected shape torch.Size([{batch_size}, {seq_len}, {num_tags}]), got {logits.shape}"
        
        print(f"Test 1 Passed: Output shape matches sequence labeling format ({batch_size}, {seq_len}, {num_tags}).")
        print("Lab task 3 passed!")
    except Exception as e:
        print(f"Test Failed: {e}")

test_task_A1()
```
